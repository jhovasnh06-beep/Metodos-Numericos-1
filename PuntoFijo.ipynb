ITERACION DE PUNTO FIJO

Queremos resolver la ecuación $x^3+4x^2-10=0$ en el intervalo $[1,2]$.
El **Teorema del Punto Fijo** establece que:

Sea $g \in C[a,b]$ tal que $g(x) \in [a,b]$ para toda $x \in [a,b]$. Además, supongamos que existe $g'(x)$ en $[a,b]$ y una constante $k < 1$ tal que:

$$|g'(x)| \leq k, \quad \forall x \in [a,b]$$

Entonces, para cualquier $p_0 \in [a,b]$, la sucesión definida por $p_n = g(p_{n-1})$ para $n \geq 1$ converge en el único punto fijo $p \in [a,b]$.

Importamos la librería numpy, que tiene funciones matemáticas, matplotlib para gráficos y pandas para mostrar tablas

```python
import numpy as np
import matplotlib.pyplot as plt
```

definimos la función $f(x)$ asociada a la ecuación $f(x)=0:$

```python
def f(x):
  return x**3 + 4*x**2 - 10.0
```

Podemos visualizar el gráfico. Estos son los hiperpárametros y los criterios de terminación del algoritmo

```python
xx=np.linspace(1,2) #genera una reticula o una partición.
plt.plot(xx,f(xx))
plt.xlabel("X")
plt.ylabel("$f(x)$")
plt.title("Gráfica $f(x)=x^3+4x^2-10$")
plt.axhline(y=0, color='black')
plt.axvline(x=0, color='black')
```

Definimos de la función de iteración g(x)

```python
def g(x):
  return np.sqrt(10 / (x + 4))
```

Evaluación de prueba para la condición inicial

```python
g(1.5)
```

```python
#Parámetros iniciales
p0 = 1.5
tol = 0.000001
nmax = 100
niter = 0

#Primera evaluación
p1 = g(p0)
error = abs(p1 - p0)

# Encabezado e impresión de la iteración 0
print("# iter\t\t p_i \t\t g(p_i) \t\t error")
print("{0} \t\t {1:6.6f} \t {2:6.6f} \t {3:6.6f}".format(niter, p0, p1, error))

# Ciclo iterativo
while error > tol and niter < nmax:
    p0 = p1 #Actualiza la aproximación
    p1 = g(p0) #Evalúa la función de iteración
    error = abs(p1 - p0) #Calcula el error absoluto
    niter += 1 #Contador de iteraciones

    # Imprime los resultados de cada iteración
    print("{0} \t\t {1:6.6f} \t {2:6.6f} \t {3:6.6f}".format(niter, p0, p1, error))

# Impresión del resultado final
print("\nLa raíz de la función obtenida por Punto Fijo es {0:6.7f}".format(p1))
```
